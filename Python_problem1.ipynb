```python
# Problem 3: 
# Splitting a bill. Four friends split a Rs 4,750 dinner.
# Print each person's share using /. Then look up 
# the // and % operators and use them to print how 
# many whole rupees each person pays and how many
# rupees are left over.

total_bill = 4750
num_friends = 4
each_share = total_bill / num_friends
print("Each person's share:", each_share)
each_share1 = total_bill // num_friends
print("Each person's share:", each_share1)
eachshare2 = total_bill % num_friends
print("Rupees left over:", eachshare2)
```

```python
# problem 4: 
# Discount and tax. A phone costs Rs 15,000. 
# Apply a 10% discount, then add 18% tax to 
# the discounted price, and print the final price. 
# Now do it the other way round: tax first, then 
# the discount. Do you get the same answer? 
# Write a comment in your code explaining why or why not.

price = 15000
discount = price * 0.10
print("discount: ", discount)
discounted_price = price - discount
print("Discounted price: ", discounted_price)
tax = discounted_price * 0.18
final_price = discounted_price + tax
print("Final price after discount and tax:", final_price)

# #for Reversing the order of operations
print("below is the calculation for tax first and then discount")
tax1 = price * 0.18
print("the result of tax1 is:", tax1)
price_with_tax = price + tax1
print("price with tax is: ", price_with_tax)
discount1 = price_with_tax * 0.10
final_price1 = price_with_tax - discount1
print("Final price after tax and discount: ", final_price1)
```

```python
# problem 5:
#  Packing boxes. A shop is shipping 23 items, and each
# box holds 6. Using only arithmetic (no if), print the 
# number of full boxes, the number of items left over, 
# and the total number of boxes needed, including a 
# partly full one (expected: 3, 5 and 4). Each box 
# costs Rs 120 to ship: print the shipping cost. 
# Test your code with 24 items. Does the box count still 
# come out right?

items = 23
box_capacity = 6
full_boxes = items // box_capacity
left_over = items % box_capacity
total_boxes = (items + box_capacity - 1) // box_capacity  # Boxes needed
shipping_cost = total_boxes * 120  # Total shipping cost
print("Full boxes are: ", full_boxes)
print("Items left over are: ", left_over)
print("Total boxes needed are: ", total_boxes)
print("Shipping cost is: ", shipping_cost)

# Leftover items can only be 0, 1, 2, 3, 4, or 5 (never 6, because 
# 6 items would make another full box).
# So the largest possible leftover is box_capacity - 1 = 5.
```
